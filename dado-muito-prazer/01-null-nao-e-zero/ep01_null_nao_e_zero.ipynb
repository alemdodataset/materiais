## Na prática: a tabela de pedidos

```
import pandas as pd

pedidos = pd.read_csv("pedidos.csv")
pedidos
```

```
# quantos vazios tem em cada coluna?
pedidos.isna().sum()
```

```
# a média ignora o vazio: (10 + 0 + 20) / 3
pedidos["desconto"].mean()
```

```
# tratando vazio como zero: (10 + 0 + 0 + 20) / 4
pedidos["desconto"].fillna(0).mean()
```

A resposta certa depende do que o vazio significa: **não teve desconto** (então é zero) ou **ninguém registrou** (então não sabemos).

```
# curiosidade: o vazio do Python (NaN) não é igual nem a ele mesmo
import numpy as np
np.nan == np.nan
```
